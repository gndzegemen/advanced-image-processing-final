```python
import cv2
import matplotlib.pyplot as plt
```

**Aşağıda BGR -> RGB dönüşümü yapmamızın sebebi matplotlib ile OpenCV kütüphanlerinin farklı renk formatlarında resmi okumasıdır.**

```python
img = cv2.imread("./nature.png")

# BGR -> RGB dönüşümü
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```
